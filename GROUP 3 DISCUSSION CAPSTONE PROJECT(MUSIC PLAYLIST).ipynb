## 2. Problem Description

A music player needs to manage a playlist that supports moving forward and backward between songs,
in addition to the usual add/search/update/delete operations. 
A **singly** linked list can only move
forward, so this project uses a **Doubly Linked List**, where every song points to both the next song
and the previous one, allowing navigation in both directions.


## 3. Objectives

By the end of this project, the system will be able to:

1. Add a song to the playlist.
2. Display all songs.
3. Search for a song by ID using **Linear Search**.
4. Update a song's information.
5. Delete a song.
6. Move to the next song.
7. Move to the previous song.
8. Count the number of songs.
9. Sort songs by title using **Bubble Sort**.
10. Sort songs by duration using **Selection Sort**.


## 4. Data Structure: Doubly Linked List

Each song is stored in a `SongNode` holding its data plus **two** references: `next` and `prev`. The
`PlaylistManager` class tracks `head`, `tail`, and a `current` pointer (the "now playing" song), which
is what makes forward/backward navigation possible.

```text
NULL <- [Song 1] <-> [Song 2] <-> [Song 3] -> NULL
                        ^
                     current
```

## 6. Class Definitions

The `SongNode` and `PlaylistManager` classes below implement the doubly linked list and all its operations.

````python
#Create empty node 
class SongNode:
    def __init__(self, song_id, title, artist, duration):
        self.song_id = song_id
        self.title = title
        self.artist = artist
        self.duration = duration   # in seconds
        self.prev = None
        self.next = None


#create playlist manager 
class PlaylistManager:
    def __init__(self):
        self.head = None
        self.tail = None
        self.current = None    # pointer for next/previous navigation

    # ---------------- ADD ----------------
    def add_song(self, song_id, title, artist, duration):
        if self._find_node(song_id):
            print(f"Error: Song ID '{song_id}' already exists.")
            return False
        new_node = SongNode(song_id, title, artist, duration)
        if not self.head:
            self.head = new_node
            self.tail = new_node
            self.current = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node
        print(f"Song '{title}' added to the playlist.")
        return True

    # ---------------- DISPLAY ----------------
    def display_songs(self):
        if not self.head:
            print("The playlist is empty.")
            return
        print(f"{'ID':<6}{'Title':<22}{'Artist':<18}{'Duration':<8}")
        print("-" * 54)
        current = self.head
        while current:
            mins, secs = divmod(current.duration, 60)
            marker = " <-- current" if current is self.current else ""
            print(f"{current.song_id:<6}{current.title:<22}{current.artist:<18}{mins}:{secs:02d}{marker}")
            current = current.next

    # ---------------- SEARCH (Linear Search) ----------------
    def _find_node(self, song_id):
        current = self.head
        while current:
            if current.song_id == song_id:
                return current
            current = current.next
        return None

    def search_song(self, song_id):
        # Linear Search: O(n)
        node = self._find_node(song_id)
        if node:
            print(f"Found: {node.song_id} | {node.title} | {node.artist} | {node.duration}s")
            return node
        print(f"Song ID '{song_id}' not found.")
        return None

    # ---------------- UPDATE ----------------
    def update_song(self, song_id, title=None, artist=None, duration=None):
        node = self._find_node(song_id)
        if not node:
            print(f"Error: Song ID '{song_id}' not found.")
            return False
        if title:
            node.title = title
        if artist:
            node.artist = artist
        if duration is not None:
            node.duration = duration
        print(f"Song '{node.title}' updated.")
        return True

    # ---------------- DELETE ----------------
    def delete_song(self, song_id):
        node = self._find_node(song_id)
        if not node:
            print(f"Error: Song ID '{song_id}' not found." if self.head else "Error: The playlist is empty.")
            return False
        if node.prev:
            node.prev.next = node.next
        else:
            self.head = node.next
        if node.next:
            node.next.prev = node.prev
        else:
            self.tail = node.prev
        if self.current is node:
            self.current = node.next or node.prev
        print(f"Song '{node.title}' deleted.")
        return True

    # ---------------- NEXT / PREVIOUS SONG ----------------
    def next_song(self):
        if not self.current:
            print("Error: The playlist is empty.")
            return None
        if self.current.next:
            self.current = self.current.next
            print(f"Now playing: {self.current.title}")
        else:
            print("Already at the last song.")
        return self.current

    def previous_song(self):
        if not self.current:
            print("Error: The playlist is empty.")
            return None
        if self.current.prev:
            self.current = self.current.prev
            print(f"Now playing: {self.current.title}")
        else:
            print("Already at the first song.")
        return self.current

    # ---------------- COUNT ----------------
    def count_songs(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    # ---------------- BUBBLE SORT (by title) ----------------
    def bubble_sort_by_title(self):
        if not self.head or not self.head.next:
            return
        swapped = True
        while swapped:
            swapped = False
            current = self.head
            while current.next:
                if current.title.lower() > current.next.title.lower():
                    self._swap_data(current, current.next)
                    swapped = True
                current = current.next
        self.current = self.head
        print("Playlist sorted alphabetically by title using Bubble Sort.")

    # ---------------- SELECTION SORT (by duration) ----------------
    def selection_sort_by_duration(self):
        if not self.head or not self.head.next:
            return
        current = self.head
        while current:
            shortest = current
            runner = current.next
            while runner:
                if runner.duration < shortest.duration:
                    shortest = runner
                runner = runner.next
            if shortest != current:
                self._swap_data(current, shortest)
            current = current.next
        self.current = self.head
        print("Playlist sorted by duration (shortest first) using Selection Sort.")

    @staticmethod
    def _swap_data(a, b):
        a.song_id, b.song_id = b.song_id, a.song_id
        a.title, b.title = b.title, a.title
        a.artist, b.artist = b.artist, a.artist
        a.duration, b.duration = b.duration, a.duration
````

## 7. Add Operation

`add_song()` attaches a new node after `tail`, linking it both forward (`tail.next`) and backward
(`new_node.prev`), then moves `tail` forward.

````python
playlist = PlaylistManager()
playlist.add_song("M001", "Sura Yako", "Sauti Sol", 220)
playlist.add_song("M002", "Up Down", "Dyce", 176)
playlist.add_song("M003", "So Sick", "Neyo", 210)
playlist.add_song("M001", "Duplicate Attempt", "Nobody", 100)  # should be rejected
playlist.display_songs()
````

## 8. Display Operation

`display_songs()` traverses from `head` to `tail`, marking whichever song `current` points to.

````python
playlist.display_songs()
````

## 9. Search Operation — Linear Search

`search_song()` checks each node's `song_id` in sequence from `head` — a **Linear Search**, O(n)
worst case.

````python
playlist.search_song("M002")   # existing song
playlist.search_song("M999")   # non-existent song
````

## 10. Update Operation

`update_song()` locates the node (via the same search logic) and updates whichever fields are given.

````python
playlist.update_song("M002", title="Up Down (Remix)")
playlist.update_song("M999", title="No Such Song")   # error case: not found
playlist.display_songs()
````

## 11. Delete Operation

`delete_song()` re-links the neighbours on **both sides** of the removed node — `node.prev.next` and
`node.next.prev` — which a singly linked list cannot do without a separate traversal to find the
predecessor. If the deleted song was `current`, the pointer moves to a neighbour so playback state
stays valid.

````python
playlist.delete_song("M003")
playlist.delete_song("M999")   # error case: not found
playlist.display_songs()
````

## 12. Sorting Operations

- **Bubble Sort** (`bubble_sort_by_title`) — repeatedly swaps adjacent out-of-order nodes until the
  playlist is sorted alphabetically by title.
- **Selection Sort** (`selection_sort_by_duration`) — repeatedly selects the shortest remaining song
  and swaps it into place.

The remaining sample songs are added first so sorting can be shown on the full playlist.

````python
for sid, title, artist, duration in [
    ("M004", "When The Party  is Over", "Billie Ellish", 260),
    ("M005", "Suzanna", "Sauti Sol", 195),
    ("M006", "Zenzele", "Uncles waffles", 240),
    ("M007", "Let Go", "Central cee", 180),
    ("M008", "Little Things", "Ell Mai", 205),
    ("M009", "Shalala", "Skeng", 190),
    ("M010", "God's Plan", "Drake", 230),
]:
    playlist.add_song(sid, title, artist, duration)

print("--- Bubble Sort by title ---")
playlist.bubble_sort_by_title()
playlist.display_songs()

print("\n--- Selection Sort by duration ---")
playlist.selection_sort_by_duration()
playlist.display_songs()

print(f"\nTotal songs: {playlist.count_songs()}")
````

## 13. Forward / Backward Navigation

This is the operation a doubly linked list makes possible that a singly linked list cannot:
`next_song()` follows `current.next`, and `previous_song()` follows `current.prev`.

````python
playlist.next_song()
playlist.next_song()
playlist.previous_song()
playlist.display_songs()
````

## 15. Complexity Analysis

| Operation | Time Complexity | Explanation |
|---|---|---|
| Add (append) | O(1) | A `tail` pointer is kept, so a new node attaches directly with no traversal |
| Display | O(n) | Visits every node once |
| Search (Linear Search) | O(n) | Worst case checks every node |
| Update | O(n) | Must first search for the node |
| Delete | O(n) | Must first search for the node; re-linking itself is O(1) once found |
| Next / Previous song | O(1) | Direct pointer follow (`current.next` / `current.prev`) — this is the main advantage of a doubly linked list here |
| Count | O(n) | One full traversal |
| Bubble Sort | O(n²) | Nested traversal comparing adjacent pairs |
| Selection Sort | O(n²) | For each position, scans remaining nodes for the shortest duration |

**Space Complexity:** O(n) for the playlist itself (each node stores one extra pointer, `prev`,
compared to a singly linked list); all operations use O(1) extra space.


## 16. Challenges

- **Keeping `current` valid after deletion:** if the currently-playing song is deleted, `current` must
  be moved to a neighbour (next, or previous if there is no next) rather than left pointing at a
  removed node.
- **Two pointers to maintain instead of one:** every insert/delete must correctly update both `next`
  and `prev` on the affected nodes — forgetting the backward link is a common bug, since forward
  traversal would still appear to work while backward traversal silently breaks.
- **Sorting a doubly linked list:** as with the singly linked list projects, both sorts swap node data
  fields rather than rewiring pointers, and `current` is reset to `head` after a sort since a song's
  position changes.
